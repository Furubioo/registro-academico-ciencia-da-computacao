```python
import numpy as np
import matplotlib.pyplot as plt

# Definindo a função y = 1 + 6 * (x**(3/2)) para a porção superior da parábola
def y(x):
    return 1 + 6 * (x**(3/2))

# Definindo os valores de x de 0 a 1
x_vals = np.linspace(0, 1, 100)

# Calculando os valores correspondentes de y
y_vals = y(x_vals)

# Criando o gráfico
plt.figure(figsize=(8, 6))
plt.plot(x_vals, y_vals, label=r'$y = 1+6x**3/2$', color='b')

# Adicionando título e legendas
plt.title('Gráfico da função 1+6x**3/2', fontsize=14)
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.grid(True)
plt.legend()

# Mostrando o gráfico
plt.show()
```

```python
import sympy as sp

# Definindo a variável simbólica
x = sp.Symbol('x')

# Função para a porção superior da curva semicúbica
y = 1 + 6 * (x**(3/2))

# Derivada de y em relação a x
dy_dx = sp.diff(y, x)

# Definindo o integrando do comprimento do arco
integrand = sp.sqrt(1 + (dy_dx)**2)

# Definindo os limites de integração
a = 0
b = 1

# Calculando a integral para obter o comprimento do arco
arc_length = sp.integrate(integrand, (x, a, b))

# Exibindo o resultado em forma simbólica e numérica
print(f"O comprimento do arco em forma simbólica é: {arc_length}")
print(f"O comprimento do arco em forma numérica é: {arc_length.evalf()}")
```
