# 💻 Prova Prática: Derivadas e Reta Tangente na Ciência da Computação

**1 Questão:** Na análise de algoritmos, muitas vezes lidamos com funções que representam o **tempo de execução** em função do tamanho da entrada.

Suponha que o tempo de execução de um algoritmo seja modelado por:


$$T(n) = n^2 + 3n + 5$$


Essa função representa um algoritmo de complexidade quadrática. Queremos entender **como o tempo de execução varia** quando o tamanho da entrada aumenta.  Qual a inclinação da curva (reta tangente) no ponto $n = 10$?


```python
n = sp.Symbol('n')
t = n**2 + 3*n + 5
t_derivada = sp.diff(t, n)

display(sp.Eq(sp.Symbol('T(n)'), t))
display(sp.Eq(sp.Symbol("T'(n)"), t_derivada))
```

```python
ponto = 10
inclinação = t_derivada.subs(n, ponto)
inclinação
```

```python
y0 = t.subs(n, ponto)
m = inclinação
reta_tangente = y0 + m*(n - ponto)
reta_tangente
```

**2 Questão:**

Considere que o tempo de execução de outro algoritmo é dado por uma das funções abaixo. Para cada uma delas:

1. Derive a função.
2. Calcule a inclinação da reta tangente no ponto \( n = 5 \).
3. Escreva a equação da reta tangente.
4. Plote o gráfico da função e da reta tangente.

### Funções sugeridas:

-  $T(n) = n^3 - 6n^2 + 12n - 5$
- $ T(n) = e^{0.5n} $



```python
n1 = sp.Symbol('n')
t1 = n**3 - 6*n**2 + 12*n - 5
t1_derivada = sp.diff(t1, n1)

display(sp.Eq(sp.Symbol('T(n)'), t1))
display(sp.Eq(sp.Symbol("T'(n)"), t1_derivada))
```

```python
ponto1 = 5
inclinação1 = t1_derivada.subs(n1, ponto)
inclinação1
```

```python
y01 = t1.subs(n1, ponto1)
m1 = inclinação1
reta_tangente1 = y01 + m1*(n1 - ponto1)
reta_tangente1
```

```python
t1_lamb = sp.lambdify(n1, t1, 'numpy')
tangente_lamb = sp.lambdify(n1, reta_tangente1, 'numpy')

n1_vals = np.linspace(3, 10)
y_vals = t1_lamb(n1_vals)
y_tan_vals = tangente_lamb(n1_vals)

plt.figure(figsize=(10,6))
plt.plot(n1_vals, y_vals, label='f(x)', color='blue')
plt.plot(n1_vals, y_tan_vals, label='Reta Tangente1 em x=5', linestyle='--', color='red')
plt.scatter([ponto1], [y01], color='black', zorder=5)
plt.title('Função e sua Reta Tangente')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(ponto, color='gray', linestyle=':')
plt.legend()
plt.show()
```

# Questão 3: Análise do Consumo de Memória de um Algoritmo

## 🧠 Contexto

Suponha que um programa de mineração de dados consome memória de acordo com a função:

$$
M(n) = n \cdot \ln(n) + 2n
$$

onde:
- $ n $ é o número de registros (em milhares) processados,
- $ M(n) $  é o consumo de memória em MB.


1. **Derive** a função $ M(n) $ para encontrar a taxa de variação do consumo de memória.
2. **Calcule a derivada no ponto $ n = 8 $**.
3. **Encontre a equação da reta tangente** no ponto \( n = 8 \).
4. **Plote** o gráfico de $ M(n) $ e da reta tangengente

```python
n3 = sp.Symbol('n')
m1 = n * sp.log(n) + 2*n
m1_derivada = sp.diff(m1, n3)

display(sp.Eq(sp.Symbol('M(n)'), m1))
display(sp.Eq(sp.Symbol("M'(n)"), m1_derivada))
```

```python
ponto3 = 8
inclinação3 = m1_derivada.subs(n3, ponto3)
inclinação3
```

```python
y03 = m1.subs(n, ponto3)
m1 = inclinação3
reta_tangente3 = y03 + m1*(n3 - ponto3)
reta_tangente3
```

```python
f_lamb = sp.lambdify(n, m1, 'numpy')
tangente_lamb = sp.lambdify(n, reta_tangente3, 'numpy')

x_vals = np.linspace(-3, 3, 400)
y_vals = f_lamb(x_vals)
y_tan_vals = tangente_lamb(x_vals)

plt.figure(figsize=(10,6))
plt.plot(x_vals, y_vals, label='f(x)', color='blue')
plt.plot(x_vals, y_tan_vals, label='Reta Tangente em x=8', linestyle='--', color='red')
plt.scatter([ponto], [y0], color='black', zorder=5)
plt.title('Função e sua Reta Tangente')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(ponto, color='gray', linestyle=':')
plt.legend()
plt.show()
```
