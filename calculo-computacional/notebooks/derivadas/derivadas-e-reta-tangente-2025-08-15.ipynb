```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import math

sp.init_printing()
```

1ª Questão:

```python
# Definir a variável simbólica x
x1 = sp.Symbol('x')
# Definir a função F(x)
y1 = 5 * x1**3 - 2*x1**2 + 7 * x1 - 4
y1_derivada = sp.diff(y1, x1)

# Exibir a função e sua derivada de forma simbólica
display(sp.Eq(sp.Symbol('F(x)'), y1))
display(sp.Eq(sp.Symbol("F'(x)"), y1_derivada))
```

```python
# Converter funções simbólicas para funções numéricas (numpy)
f_lamb = sp.lambdify(x1, y1, 'numpy')
f_deriv_lamb = sp.lambdify(x1, y1_derivada, 'numpy')

# Definir valores de x para plot
x_vals = np.linspace(-3, 3, 400)
y_vals = f_lamb(x_vals)
y_deriv_vals = f_deriv_lamb(x_vals)

# Plotar a função e sua derivada
plt.figure(figsize=(10, 6))
plt.plot(x_vals, y_vals, label='f(x)', color='blue')  # Função original
plt.plot(x_vals, y_deriv_vals, label="f'(x)", linestyle='--', color='red') # Derivada
plt.title('Função e sua Derivada')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axhline(0, color='black', lw=0.5, linestyle='--')
plt.axvline(0, color='black', lw=0.5, linestyle='--')
plt.legend()
plt.show()
```

Interpretação dos pontos críticos:

```python
# Variável simbólica
x14 = sp.Symbol('x')

# Função cúbica
f14 = x14**3 - 3*x14**2 - 9*x14 + 1

# Derivada simbólica
f14_derivada = sp.diff(f14, x14)

# Exibir função e derivada
display(sp.Eq(sp.Symbol('f(x)'), f14))
display(sp.Eq(sp.Symbol("f'(x)"), f14_derivada))

pontos_criticos = sp.solve(sp.Eq(f14_derivada, 0), x14)
print("Pontos críticos:", pontos_criticos)
```

```python
# Converter para funções numéricas e plotar
f_lamb = sp.lambdify(x14, f14, 'numpy')
f_deriv_lamb = sp.lambdify(x14, f14_derivada, 'numpy')
x_vals = np.linspace(-5, 5, 400)
y_vals = f_lamb(x_vals)
y_deriv_vals = f_deriv_lamb(x_vals)

plt.figure(figsize=(10,6))
plt.plot(x_vals, y_vals, label='f(x)', color='blue')
plt.plot(x_vals, y_deriv_vals, label="f'(x)", linestyle='--', color='red')
# Marcar pontos críticos
for pc in pontos_criticos:
    plt.scatter(float(pc), f_lamb(pc), color='black', zorder=5)
plt.title('Função, Derivada e Pontos Críticos')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```
