# 💻 Prova Prática: Derivadas e Reta Tangente na Ciência da Computação

## 🎯 Objetivos:

- Derivar funções que modelam tempo de execução de algoritmos.
- Determinar a taxa de crescimento (inclinação da reta tangente) em pontos específicos.
- Visualizar o comportamento da função e sua reta tangente usando gráficos.

---



**1 Questão:** Na análise de algoritmos, muitas vezes lidamos com funções que representam o **tempo de execução** em função do tamanho da entrada.

Suponha que o tempo de execução de um algoritmo seja modelado por:


$$T(n) = n^2 + 3n + 5$$


Essa função representa um algoritmo de complexidade quadrática. Queremos entender **como o tempo de execução varia** quando o tamanho da entrada aumenta.  Qual a inclinação da curva (reta tangente) no ponto $n = 10$?


1a Questão:

```python
n = sp.Symbol('n')
T = n**2 + 3*n + 5
T_derivada = sp.diff(T, n)

display(sp.Eq(sp.Symbol('T(n)'), T))
display(sp.Eq(sp.Symbol("T'(n)"), T_derivada))
```

```python
ponto = 10
inclinação = T_derivada.subs(n, ponto)
inclinação
```

```python
y0 = T.subs(n, ponto)
m = inclinação
reta_tangente = y0 + m*(n - ponto)
reta_tangente
```

2a Questão:

```python
n1 = sp.Symbol('n')
T1 = n**3 - 6*n**2 + 12*n - 5
T1_derivada = sp.diff(T1, n1)

display(sp.Eq(sp.Symbol('T(n)'), T1))
display(sp.Eq(sp.Symbol("T'(n)"), T1_derivada))
```

```python
ponto1 = 5
inclinação1 = T1_derivada.subs(n1, ponto1)
inclinação1
```

```python
y0 = T1.subs(n1, ponto1)
m = inclinação
reta_tangente = y0 + m*(n1 - ponto1)
reta_tangente
```

```python
f_lamb = sp.lambdify(n1, T1, 'numpy')
tangente_lamb = sp.lambdify(n1, reta_tangente, 'numpy')

x_vals = np.linspace(3, 7)
y_vals = f_lamb(x_vals)
y_tan_vals = tangente_lamb(x_vals)

plt.figure(figsize=(10,6))
plt.plot(x_vals, y_vals, label='f(x)', color='blue')
plt.plot(x_vals, y_tan_vals, label='Reta Tangente em x=1', linestyle='--', color='red')
plt.scatter([ponto1], [y0], color='black', zorder=5)
plt.title('Função e sua Reta Tangente')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(ponto, color='gray', linestyle=':')
plt.legend()
plt.show()
```

# Questão 3: Análise do Consumo de Memória de um Algoritmo

## 🧠 Contexto

Suponha que um programa de mineração de dados consome memória de acordo com a função:

$$
M(n) = n \cdot \ln(n) + 2n
$$

onde:
- $ n $ é o número de registros (em milhares) processados,
- $ M(n) $  é o consumo de memória em MB.


1. **Derive** a função $ M(n) $ para encontrar a taxa de variação do consumo de memória.
2. **Calcule a derivada no ponto $ n = 8 $**.
3. **Encontre a equação da reta tangente** no ponto \( n = 8 \).
4. **Plote** o gráfico de $ M(n) $ e da reta tangengente

3a Questão:

```python
n3 = sp.Symbol('n')
M = n* sp.ln(n3) + 2 * n
M_derivada = sp.diff(M, n3)

display(sp.Eq(sp.Symbol('M(n)'), M))
display(sp.Eq(sp.Symbol("M'(n)"), M_derivada))
```

```python
ponto3 = 8
inclinação3 = M_derivada.subs(n3, ponto3)
inclinação3
```

```python
y3 = M.subs(n3, ponto3)
m3 = inclinação3
reta_tangente3 = y3 + m3*(n3 - ponto3)
reta_tangente3
```

```python
f_lamb = sp.lambdify(n3, M, 'numpy')
tangente_lamb = sp.lambdify(n3, reta_tangente3, 'numpy')

x_vals = np.linspace(-2,16)
y_vals = f_lamb(x_vals)
y_tan_vals = tangente_lamb(x_vals)

plt.figure(figsize=(10,6))
plt.plot(x_vals, y_vals, label='M(n)', color='blue')
plt.plot(x_vals, y_tan_vals, label='Reta Tangente em x=8', linestyle='--', color='red')
plt.scatter([ponto3], [y3], color='black', zorder=5)
plt.title('Função e sua Reta Tangente')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.axvline(ponto, color='gray', linestyle=':')
plt.legend()
plt.show()
```
