 # **Aplicações de Integrais em ciênça da computação**



### 1. **Algoritmos de Aprendizado de Máquina (Machine Learning)**
   - **Exemplo:** No aprendizado de máquina, particularmente na inferência bayesiana, usamos integrais para calcular a probabilidade de uma hipótese dado um conjunto de dados (a probabilidade posterior). \\

   $   P(\theta | D) = \frac{P(D | \theta) P(\theta)}{\int P(D | \theta') P(\theta') d\theta'}
   $

   Aqui, a integral no denominador é usada para normalizar a distribuição, garantindo que a soma das probabilidades seja 1.

### 2. **Análise de Algoritmos**
   - **Exemplo:** Em análise de algoritmos, ao calcular a complexidade de um algoritmo de ordenação como o QuickSort, a integral pode ser usada para determinar o tempo médio de execução. Para analisar o caso médio, a integral é usada para calcular o número esperado de comparações:
   
   $T(n) = \frac{2}{n} \int_1^n T(k)  $dk$
   
   onde $T(n)$ é o tempo esperado de execução para ordenar uma lista de tamanho $n$.

### 3. **Computação Gráfica**
   - **Exemplo:** Na computação gráfica, para calcular a iluminação global de uma cena, podemos usar o método de Monte Carlo para aproximar a integral de radiância que atinge um ponto em uma superfície:
  
  $   L(x, \omega) = \int_{\Omega} f_r(x, \omega', \omega) L(x, \omega') (\omega' \cdot n) \, d\omega'$

   onde $L(x, \omega)$ é a radiância na direção $\omega$ de um ponto $x$, e a integral é sobre todas as direções $\omega'$ de onde a luz pode vir.

### 4. **Visão Computacional**
   - **Exemplo:** Na filtragem de imagens, a convolução é usada para aplicar filtros que destacam certas características da imagem, como bordas. A convolução contínua de uma imagem $I(x, y)$ com um filtro $h(x, y)$ é:
   
   $(I * h)(x, y) = \int_{-\infty}^{\infty} \int_{-\infty}^{\infty} I(u, v) h(x - u, y - v) \, du \, dv$

   Essa integral calcula o valor de um pixel filtrado com base na média ponderada de seus vizinhos.

### 5. **Teoria da Informação**
   - **Exemplo:** Para variáveis contínuas, a entropia diferencial é uma extensão da entropia de Shannon e é calculada usando uma integral:
   
   $H(X) = - \int_{-\infty}^{\infty} p(x) \log p(x) \, dx$

   onde $p(x)$ é a função densidade de probabilidade da variável aleatória $X$.

### 6. **Sistemas Dinâmicos e Controle**
   - **Exemplo:** Em controle de sistemas, o controle proporcional-integral-derivativo (PID) usa uma integral para calcular o controle proporcional ao erro acumulado ao longo do tempo:
   
   $u(t) = K_p e(t) + K_i \int_0^t e(\tau) \, d\tau + K_d \frac{de(t)}{dt}$

   onde $e(t)$ é o erro no tempo $t$, e $K_p$, $K_i$, e $K_d$ são constantes de ganho.

### 7. **Simulação e Modelagem Matemática**
   - **Exemplo:** Em simulação de sistemas físicos, uma equação diferencial pode modelar o movimento de um objeto. A integral da função de força $F(t)$ sobre o tempo dá a mudança no momento linear de um objeto:
   
   $p(t) = \int_0^t F(\tau) \, d\tau$
   
   onde $p(t)$ é o momento linear no tempo $t$.

### 8. **Teoria das Redes e Grafos**
   - **Exemplo:** Em redes de comunicação, para modelar o fluxo de dados contínuos, como pacotes de rede em um link, a integral é usada para calcular a quantidade acumulada de dados transferidos em um intervalo de tempo:
   
   $Q(t) = \int_{0}^{t} R(\tau) \, d\tau$
   
   onde $R(t)$ é a taxa de transferência de dados no tempo $t$.

Esses exemplos mostram como as integrais são aplicadas de maneira prática em diferentes áreas da ciência da computação.

### **Exemplo Simples de Inferência Bayesiana**
Vamos considerar um exemplo de inferência bayesiana onde usamos uma função exponencial simples para a verossimilhança e aplicamos a regra de substituição para resolver a integral.

#### **Contexto do Problema**

Imagine que estamos observando o tempo entre eventos (como o tempo até que uma pessoa chegue a uma fila de atendimento), e acreditamos que esse tempo segue uma distribuição exponencial. Queremos estimar o parâmetro $\lambda$ (taxa de eventos) dessa distribuição com base em uma única observação.

#### **Definições:**

- **Dados Observados (D):** Temos um único tempo observado $x = 2$ minutos.
- **Hipótese:** Queremos estimar $\lambda$, que é o parâmetro da distribuição exponencial $X \sim \text{Exp}(\lambda)$.
- **Prior ($P(\lambda)$):** Suponhamos que nossa crença inicial sobre $\lambda$ seja uma distribuição uniforme entre 0 e 3:
  $
  P(\lambda) = \frac{1}{3}, \quad \text{para } \lambda \in [0, 3].
  $
- **Likelihood ($P(x | \lambda)$):** A verossimilhança de observar $x = 2$ dado $\lambda$ é dada pela função densidade da distribuição exponencial:
  $
  P(x | \lambda) = \lambda e^{-\lambda x}.
  $

#### **Passo 1: Calculando a Posterior**

Queremos encontrar a probabilidade posterior de $\lambda$ dado $x = 2$:

$
P(\lambda | x) = \frac{P(x | \lambda) P(\lambda)}{P(x)}.
$

Para calcular essa probabilidade, precisamos encontrar $P(x)$, a probabilidade total dos dados. Isso é feito pela integral:

$
P(x) = \int_{0}^{3} P(x | \lambda) P(\lambda) \, d\lambda.
$

Substituindo os valores:

$
P(x) = \int_{0}^{3} \left( \lambda e^{-\lambda \cdot 2} \right) \left( \frac{1}{3} \right) \, d\lambda.
$

Simplificando a constante:

$
P(x) = \frac{1}{3} \int_{0}^{3} \lambda e^{-2\lambda} \, d\lambda.
$

### **Integral a Resolver**

A integral que queremos resolver é:

$
\int_{0}^{3} \lambda e^{-2\lambda} \, d\lambda.
$

### **Passo 1: Substituição de Variáveis**

Vamos fazer a substituição:

$
u = -2\lambda \implies \frac{du}{d\lambda} = -2 \implies d\lambda = -\frac{1}{2} \, du.
$

Quando $\lambda = 0$, então $u = -2 \cdot 0 = 0$.

Quando $\lambda = 3$, então $u = -2 \cdot 3 = -6$.

Substituindo $\lambda$ e $d\lambda$ na integral, obtemos:

$
\int_{0}^{3} \lambda e^{-2\lambda} \, d\lambda = \int_{0}^{-6} \left( -\frac{u}{2} \right) e^{u} \left( -\frac{1}{2} \right) \, du.
$

Simplificando, a integral se torna:

$
= \frac{1}{4} \int_{0}^{-6} u e^{u} \, du.
$

### **Passo 2: Integral por Partes**

Para resolver $\int u e^u \, du$, aplicamos a integração por partes. Seja:

$
v = u, \quad dv = du, \quad dw = e^u \, du, \quad w = e^u.
$

Então, a integral por partes é:

$
\int u e^u \, du = u e^u - \int e^u \, du = u e^u - e^u + C.
$

Aplicando os limites de integração de $0$ a $-6$:

$
\left[ u e^u - e^u \right]_{0}^{-6}.
$

Substituindo os limites:

$
\left[ (-6) e^{-6} - e^{-6} \right] - \left[ (0) e^0 - e^0 \right].
$

Simplificando:

$
(-6e^{-6} - e^{-6}) - (-1) = -7e^{-6} + 1.
$

### **Passo 3: Multiplicando pelo Coeficiente Externo**

Voltando à integral original, temos:

$
\frac{1}{4} \left(-7e^{-6} + 1\right).
$

### **Resultado Final da Integral**

Portanto, a solução da integral é:

$
\int_{0}^{3} \lambda e^{-2\lambda} \, d\lambda = \frac{1}{4} \left(1 - 7e^{-6}\right).
$

Esta é a resposta da integral resolvida com substituição e integração por partes.


# Agora utilizando Código

```python
import sympy as sp

# Definir a variável
λ = sp.Symbol('λ')

# Definir as funções de likelihood e prior
P_x_given_lambda = λ * sp.exp(-2 * λ)  # Função de verossimilhança
P_lambda = 1/3  # Prior uniforme entre 0 e 3

# Definir a integral a ser resolvida
integral_expr = P_lambda * P_x_given_lambda
integral_result = sp.integrate(integral_expr, (λ, 0, 3))

integral_result
```

```python
import numpy as np
import matplotlib.pyplot as plt


posterior_density = sp.lambdify(λ, integral_expr, "numpy")
# Criar um intervalo de valores para λ
λ_vals = np.linspace(0, 3, 500)
posterior_vals = posterior_density(λ_vals)

# Plotar a densidade da função posterior
plt.figure(figsize=(8, 5))
plt.plot(λ_vals, posterior_vals, label=r'Posterior Density $\lambda e^{-2\lambda} \cdot \frac{1}{3}$', color='blue')
plt.title('Densidade da Função Posterior')
plt.xlabel(r'$\lambda$')
plt.ylabel(r'$P(x | \lambda) P(\lambda)$')
plt.legend()
plt.grid(True)
plt.show()

# Análise da Integral e do Resultado
result_integral = 0.0833 - 0.5833 * np.exp(-6)

# Gráfico com o resultado da integral
plt.figure(figsize=(8, 5))
plt.plot(λ_vals, posterior_vals, label=r'Posterior Density $\lambda e^{-2\lambda} \cdot \frac{1}{3}$', color='blue')
plt.axhline(y=result_integral, color='red', linestyle='--', label=f'Resultado da Integral: {result_integral:.4f}')
plt.title('Densidade da Função Posterior com Resultado da Integral')
plt.xlabel(r'$\lambda$')
plt.ylabel(r'$P(x | \lambda) P(\lambda)$')
plt.legend()
plt.grid(True)
plt.show()
```

### **Análise de Algoritmo: QuickSort e Complexidade Média**

O QuickSort é um algoritmo de ordenação eficiente que usa a estratégia de "dividir e conquistar" para ordenar uma lista. A ideia básica é escolher um elemento da lista como "pivô", particionar os outros elementos em torno do pivô, e então aplicar recursivamente o mesmo processo às sub-listas geradas.

#### **Objetivo: Determinar o Tempo Médio de Execução**

Para analisar o tempo médio de execução do QuickSort, assumimos que o pivô é escolhido aleatoriamente. O tempo esperado de execução, $ T(n) $, pode ser determinado pela integral que representa o número esperado de comparações para uma lista de tamanho $ n $:

$
T(n) = \frac{2}{n} \int_{1}^{n} T(k) \, dk.
$

Neste caso:

- $ T(n) $ é o tempo médio de execução para ordenar uma lista de tamanho $ n $.
- A integral representa o custo médio de dividir a lista de tamanho $ n $ em sub-listas de tamanhos variáveis $ k $.

Vamos considerar um exemplo mais complexo para analisar o tempo médio de execução do QuickSort, utilizando uma função mais complicada para $T(k)$.

Vamos considerar a seguinte função para o número de comparações:

$
T(k) = k \cdot \ln(k) \cdot e^{-k}.
$

Queremos calcular o tempo médio de execução $ T(n) $ para o QuickSort em termos de uma integral:

$
T(n) = \frac{2}{n} \int_{1}^{2} k^5 \ln(k) \, dk.
$




### **Resolução Manual da Integral:**

Para resolver a integral manualmente, podemos usar a técnica de **integração por partes**. A integração por partes é aplicada quando temos o produto de duas funções que podem ser diferenciadas e integradas separadamente.

#### **Definindo a Integração por Partes**

Escolhemos:

$
u = \ln(k), \quad dv = k^5 \, dk.
$

Agora, derivamos $u$ e integramos $dv$:

$
du = \frac{1}{k} \, dk, \quad v = \frac{k^6}{6}.
$

A fórmula da integração por partes é:

$
\int u \, dv = uv - \int v \, du.
$

Substituindo os valores:

$
\int_{1}^{2} k^5 \ln(k) \, dk = \left[ \ln(k) \cdot \frac{k^6}{6} \right]_{1}^{2} - \int_{1}^{2} \frac{k^6}{6} \cdot \frac{1}{k} \, dk.
$

Simplificando a integral restante:

$
= \left[ \frac{k^6 \ln(k)}{6} \right]_{1}^{2} - \frac{1}{6} \int_{1}^{2} k^5 \, dk.
$

#### **Resolver a Integral Restante**

A integral restante é uma simples integral polinomial:

$
\int_{1}^{2} k^5 \, dk = \left[ \frac{k^6}{6} \right]_{1}^{2} = \frac{2^6}{6} - \frac{1^6}{6} = \frac{64}{6} - \frac{1}{6} = \frac{63}{6} = 10.5.
$

#### **Substituir e Calcular o Valor Final**

Substituindo de volta na expressão:

$
\int_{1}^{2} k^5 \ln(k) \, dk = \left( \frac{2^6 \ln(2)}{6} - \frac{1^6 \ln(1)}{6} \right) - \frac{1}{6} \cdot 10.5.
$

Como $\ln(1) = 0$, temos:

$
= \frac{64 \ln(2)}{6} - \frac{10.5}{6}.
$

Simplificando:

$
= \frac{64 \ln(2) - 10.5}{6}.
$

Multiplicando pela constante fora da integral, temos:

$
T(n) = \frac{2}{n} \left( \frac{64 \ln(2) - 10.5}{6} \right).
$


```python
# Reimportar bibliotecas após a reinicialização do estado
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Definir a variável simbólica
k = sp.Symbol('k')

# Definir a função a ser integrada
integrand = k**5 * sp.ln(k)

# Calcular a integral no intervalo de 1 a 2
integral_result = sp.integrate(integrand, (k, 1, 2))

# Definir n como uma variável simbólica para a fórmula de T(n)
n = sp.Symbol('n')
T_n_expr = (2/n) * integral_result
print("Integral de f(x) =", T_n_expr)
# Converter a expressão simbólica de T(n) em uma função numérica para plotagem
T_n_func = sp.lambdify(n, T_n_expr, 'numpy')

# Gerar valores para n para a análise gráfica
n_vals = np.linspace(1.1, 10, 100)  # Começando de 1.1 para evitar divisão por zero
T_n_vals = T_n_func(n_vals)

# Plotar o resultado
plt.figure(figsize=(8, 5))
plt.plot(n_vals, T_n_vals, label=r'$T(n) = \frac{2}{n} \int_{1}^{2} k^5 \ln(k) \, dk$', color='blue')
plt.title('Análise Gráfica da Função $T(n)$')
plt.xlabel('n')
plt.ylabel('T(n)')
plt.grid(True)
plt.legend()
plt.show()
```

1. $P(x) = \int_{0}^{3} \lambda^2 e^{-λ}\, d\lambda.$


```python
# Definir a variável
λ = sp.Symbol('λ')

# Definir as funções de likelihood e prior
P_x_given_lambda = λ**2 * sp.exp(-λ)  # Função de verossimilhança
# P_lambda = 1/3  # Prior uniforme entre 0 e 3

# Definir a integral a ser resolvida
integral_expr = P_x_given_lambda
integral_result = sp.integrate(integral_expr, (λ, 0, 3))

integral_result
```

```python
# Definir a variável
k = sp.Symbol('k')

# Definir as funções de likelihood e prior
P_x_given_lambda =  sp.exp(k) * sp.cos(k) # Função de verossimilhança
P_lambda = 2 / sp.pi  # Prior uniforme entre 2 e pi

# Definir a integral a ser resolvida
integral_expr = P_lambda * P_x_given_lambda
integral_result = sp.integrate(integral_expr, (k, 1, n))

integral_result
```

```python
posterior_density = sp.lambdify(k, integral_expr, "numpy")
# Criar um intervalo de valores para λ
k_vals = np.linspace(0, 3, 500)
posterior_vals = posterior_density(k_vals)

# Plotar a densidade da função posterior
plt.figure(figsize=(8, 5))
plt.plot(k_vals, posterior_vals, label=f'{integral_expr}', color='blue')
plt.title('Densidade da Função Posterior')
plt.xlabel(r'$\lambda$')
plt.ylabel(r'$P(x | \lambda) P(\lambda)$')
plt.legend()
plt.grid(True)
plt.show()

# Análise da Integral e do Resultado
result_integral = integral_result.subs(n, 3).evalf()
result_integral = float(result_integral)

# Gráfico com o resultado da integral
plt.figure(figsize=(8, 5))
plt.plot(k_vals, posterior_vals, label=f'{integral_expr}', color='blue')
plt.axhline(y=result_integral, color='red', linestyle='--',  label=f'Resultado da Integral: {result_integral}')
plt.title('Densidade da Função Posterior com Resultado da Integral')
plt.xlabel(r'$\lambda$')
plt.ylabel(r'$P(x | \lambda) P(\lambda)$')
plt.legend()
plt.grid(True)
plt.show()
```
