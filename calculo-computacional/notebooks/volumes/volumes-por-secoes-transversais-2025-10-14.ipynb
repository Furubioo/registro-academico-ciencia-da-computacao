```python
import sympy as sp

# Definição das variáveis
x = sp.symbols('x')
r = sp.symbols('r', positive=True)  # Raio da esfera

# Função para a área da seção transversal A(x) = π * (r² - x²)
A_x = sp.pi * (r**2 - x**2)

# Cálculo do volume usando integração definida de -r até r
volume = sp.integrate(A_x, (x, -r, r))
volume
# Simplificação do resultado
volume_simplificado = sp.simplify(volume)
print(f"O volume da esfera é: {volume}")
print(f"O volume da esfera é: {volume_simplificado}")
```
