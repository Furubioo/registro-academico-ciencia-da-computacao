```python
def plot_function(f_sym, x_min, x_max, y_min=None, y_max=None, num_points=1000):
    x = sp.symbols('x')
    f_lambda = sp.lambdify(x, f_sym, 'numpy')  # Converte a função sympy para numpy

    x_vals = np.linspace(x_min, x_max, num_points)  # Gera pontos no intervalo especificado
    y_vals = f_lambda(x_vals)  # Avalia a função nos pontos gerados

    plt.figure(figsize=(8, 5))
    plt.plot(x_vals, y_vals, label=f"y = {f_sym}", color='b')
    plt.axhline(0, color='black', linewidth=0.5)
    plt.axvline(0, color='black', linewidth=0.5)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.legend()
    plt.xlabel("x")
    plt.ylabel("g(x)")
    plt.title("Gráfico da Função")

    if y_min is not None and y_max is not None:
        plt.ylim(y_min, y_max)  # Define o intervalo do eixo y

    plt.show()
```

```python
x = sp.symbols('x')
g = (x**2 - x + 6) / (x - 2)
plot_function(g, -3, 6, y_min=-50, y_max=50)
```

```python
def is_continuous(f_sym, x_val):
    x = sp.symbols('x')
    left_limit = sp.limit(f_sym, x, x_val, dir='-')
    right_limit = sp.limit(f_sym, x, x_val, dir='+')
    function_value = f_sym.subs(x, x_val)

    print(f"Limite pela esquerda em x = {x_val}: {left_limit}")
    print(f"Limite pela direita em x = {x_val}: {right_limit}")
    print(f"Valor da função em x = {x_val}: {function_value}")

    return left_limit == right_limit == function_value
```

```python
x_val = 2
print(f"A função é contínua em x = {x_val}?", is_continuous(g,x_val))
```
