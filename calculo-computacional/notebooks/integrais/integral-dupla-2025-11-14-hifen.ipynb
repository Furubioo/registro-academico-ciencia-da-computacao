### Exemplo Teórico 1: Integral Dupla
**Problema**: Calcule a integral dupla da função $ f(x, y) = x^2 + y^2 $ sobre a região retangular onde $ 0 \leq x \leq 1 $ e $ 0 \leq y \leq 2 $.

### Passo 1: Definir a integral
A integral a ser resolvida é:
$$
\iint_R (x^2 + y^2) \, dx\,dy = \int_{0}^{2} \int_{0}^{1} (x^2 + y^2) \, dx\,dy
$$

### Passo 2: Resolver a integral interna ($\int_{0}^{1} (x^2 + y^2) \, dx$)
Vamos resolver a integral em relação a $ x $ primeiro, tratando $ y $ como uma constante:
$$
\int_{0}^{1} (x^2 + y^2) \, dx = \int_{0}^{1} x^2 \, dx + \int_{0}^{1} y^2 \, dx
$$

- **Integral de $ x^2 $**:
$$
\int_{0}^{1} x^2 \, dx = \frac{x^3}{3} \Big|_{0}^{1} = \frac{1}{3}
$$

- **Integral de $ y^2 $**:
$$
\int_{0}^{1} y^2 \, dx = y^2 \cdot x \Big|_{0}^{1} = y^2 \cdot 1 = y^2
$$

Portanto:
$$
\int_{0}^{1} (x^2 + y^2) \, dx = \frac{1}{3} + y^2
$$

### Passo 3: Resolver a integral externa ($\int_{0}^{2} (\frac{1}{3} + y^2) \, dy$)
Agora, vamos resolver a integral em relação a $ y $:
$$
\int_{0}^{2} \left(\frac{1}{3} + y^2\right) \, dy = \int_{0}^{2} \frac{1}{3} \, dy + \int_{0}^{2} y^2 \, dy
$$

- **Integral de $ \frac{1}{3} $**:
$$
\int_{0}^{2} \frac{1}{3} \, dy = \frac{1}{3} \cdot y \Big|_{0}^{2} = \frac{1}{3} \cdot 2 = \frac{2}{3}
$$

- **Integral de $ y^2 $**:
$$
\int_{0}^{2} y^2 \, dy = \frac{y^3}{3} \Big|_{0}^{2} = \frac{2^3}{3} = \frac{8}{3}
$$

Portanto:
$$
\int_{0}^{2} \left(\frac{1}{3} + y^2\right) \, dy = \frac{2}{3} + \frac{8}{3} = \frac{10}{3}
$$

### Resultado Final
$$
\iint_R (x^2 + y^2) \, dx\,dy = \frac{10}{3}
$$


```python
import sympy as sp
from sympy import integrate, symbols

# Definindo as variáveis simbólicas
x, y = symbols('x y')

# Definindo a função a ser integrada
f = x**2 + y**2

# Calculando a integral interna em relação a x
integral_interna = integrate(f, (x, 0, 1))

# Calculando a integral externa em relação a y
integral_resultado = integrate(integral_interna, (y, 0, 2))

# Exibindo o resultado
print('Resultado da integral dupla:', integral_resultado)
```

### Exemplo Teórico 2: Integral Dupla sobre uma Região Triangular
**Problema**: Calcule a integral dupla da função $ f(x, y) = xy $ sobre a região triangular delimitada por $ 0 \leq x \leq 2 $ e $ 0 \leq y \leq x $.

**Interpretação**:
Esta integral dupla calcula a soma ponderada dos valores de $ xy $ sobre a região triangular no plano $ xy $ definida pelos limites $ x $ de 0 a 2 e $ y $ de 0 a $ x $.

### Solução Passo a Passo
1. Defina a integral dupla:
$$
\iint_R xy \, dy\,dx = \int_{0}^{2} \int_{0}^{x} xy \, dy\,dx
$$

2. Resolva a integral interna ($\int_{0}^{x} xy \, dy$):
$$
\int_{0}^{x} xy \, dy = x \int_{0}^{x} y \, dy = x \cdot \frac{y^2}{2} \Big|_{0}^{x} = x \cdot \frac{x^2}{2} = \frac{x^3}{2}
$$

3. Resolva a integral externa ($\int_{0}^{2} \frac{x^3}{2} \, dx$):
$$
\int_{0}^{2} \frac{x^3}{2} \, dx = \frac{1}{2} \int_{0}^{2} x^3 \, dx = \frac{1}{2} \cdot \frac{x^4}{4} \Big|_{0}^{2} = \frac{1}{2} \cdot \frac{16}{4} = 2
$$

### Resultado Final:
$$
\iint_R xy \, dy\,dx = 2
$$

### Implementação em Python:




```python
import sympy as sp
from sympy import integrate

# Definindo as variáveis simbólicas
x, y = symbols('x y')

# Definindo a função a ser integrada
f = x * y

# Calculando a integral interna (em relação a y)
integral_interna = integrate(f, (y, 0, x))

# Calculando a integral externa (em relação a x)
integral_resultado = integrate(integral_interna, (x, 0, 2))

# Exibindo o resultado
print('Resultado da integral dupla sobre a região triangular:', integral_resultado)
```

### Exemplo Teórico: Integral Tripla em uma Região Cúbica
**Problema**: Calcule a integral tripla da função $ f(x, y, z) = x + y + z $ sobre a região cúbica onde $ 0 \leq x \leq 1 $, $ 0 \leq y \leq 1 $ e $ 0 \leq z \leq 1 $.

### Solução Passo a Passo:
1. **Defina a integral tripla**:
$$
\iiint_V (x + y + z) \, dx\,dy\,dz = \int_{0}^{1} \int_{0}^{1} \int_{0}^{1} (x + y + z) \, dx\,dy\,dz
$$

2. **Resolva a integral interna ($\int_{0}^{1} (x + y + z) \, dx$)**:
$$
\int_{0}^{1} (x + y + z) \, dx = \int_{0}^{1} x \, dx + \int_{0}^{1} y \, dx + \int_{0}^{1} z \, dx
$$

Como $ y $ e $ z $ são constantes em relação a $ x $:
- $\int_{0}^{1} x \, dx = \frac{x^2}{2} \Big|_{0}^{1} = \frac{1}{2}$
- $\int_{0}^{1} y \, dx = y \cdot x \Big|_{0}^{1} = y$
- $\int_{0}^{1} z \, dx = z \cdot x \Big|_{0}^{1} = z$

Portanto:
$$
\int_{0}^{1} (x + y + z) \, dx = \frac{1}{2} + y + z
$$

3. **Resolva a integral em relação a $ y $ ($\int_{0}^{1} (\frac{1}{2} + y + z) \, dy$)**:
$$
\int_{0}^{1} \left(\frac{1}{2} + y + z\right) \, dy = \int_{0}^{1} \frac{1}{2} \, dy + \int_{0}^{1} y \, dy + \int_{0}^{1} z \, dy
$$

- $\int_{0}^{1} \frac{1}{2} \, dy = \frac{1}{2} \cdot y \Big|_{0}^{1} = \frac{1}{2}$
- $\int_{0}^{1} y \, dy = \frac{y^2}{2} \Big|_{0}^{1} = \frac{1}{2}$
- $\int_{0}^{1} z \, dy = z \cdot y \Big|_{0}^{1} = z$

Portanto:
$$
\int_{0}^{1} \left(\frac{1}{2} + y + z\right) \, dy = \frac{1}{2} + \frac{1}{2} + z = 1 + z
$$

4. **Resolva a integral em relação a $ z $ ($\int_{0}^{1} (1 + z) \, dz$)**:
$$
\int_{0}^{1} (1 + z) \, dz = \int_{0}^{1} 1 \, dz + \int_{0}^{1} z \, dz
$$

- $\int_{0}^{1} 1 \, dz = z \Big|_{0}^{1} = 1$
- $\int_{0}^{1} z \, dz = \frac{z^2}{2} \Big|_{0}^{1} = \frac{1}{2}$

Portanto:
$$
\int_{0}^{1} (1 + z) \, dz = 1 + \frac{1}{2} = \frac{3}{2}
$$

### Resultado Final:
$$
\iiint_V (x + y + z) \, dx\,dy\,dz = \frac{3}{2}
$$



```python
from sympy import symbols, integrate

# Definindo as variáveis simbólicas
x, y, z = symbols('x y z')

# Definindo a função a ser integrada
f = x + y + z

# Calculando a integral interna (em relação a x)
integral_interna = integrate(f, (x, 0, 1))

# Calculando a integral intermediária (em relação a y)
integral_intermediaria = integrate(integral_interna, (y, 0, 1))

# Calculando a integral externa (em relação a z)
integral_resultado = integrate(integral_intermediaria, (z, 0, 1))

# Exibindo o resultado
print('Resultado da integral tripla:', integral_resultado)
```


### Exemplo Teórico: Integral Tripla em Coordenadas Cilíndricas
**Problema**: Calcule o volume de um cone com altura $ h = 2 $ e raio da base $ R = 2 $ usando uma integral tripla em coordenadas cilíndricas.

### Solução Teórica:
Em coordenadas cilíndricas, uma função $ f(x, y, z) $ é representada por $ f(r, \theta, z) $, onde:
- $ x = r \cos(\theta) $
- $ y = r \sin(\theta) $
- O jacobiano de transformação é $ r $, então $ dV = r \, dr\,d\theta\,dz $

O volume do cone pode ser calculado por:
$$
\iiint_V 1 \cdot r \, dr\,d\theta\,dz
$$

Os limites de integração são:
- $ \theta $ de $ 0 $ a $ 2\pi $
- $ z $ de $ 0 $ a $ h $ (onde $ h = 2 $)
- $ r $ de $ 0 $ a $ \frac{R}{h}z $ (o raio varia linearmente de 0 na base até $ R $ no topo)

### Resolvendo a Integral:
$$
\int_{0}^{2} \int_{0}^{2\pi} \int_{0}^{\frac{2}{2}z} r \, dr\,d\theta\,dz
$$

1. **Integral em relação a $ r $**:
$$
\int_{0}^{\frac{2}{2}z} r \, dr = \frac{r^2}{2} \Big|_{0}^{z} = \frac{z^2}{2}
$$

2. **Integral em relação a $ \theta $**:
$$
\int_{0}^{2\pi} \frac{z^2}{2} \, d\theta = \frac{z^2}{2} \cdot 2\pi = \pi z^2
$$

3. **Integral em relação a $ z $**:
$$
\int_{0}^{2} \pi z^2 \, dz = \pi \cdot \frac{z^3}{3} \Big|_{0}^{2} = \pi \cdot \frac{8}{3} = \frac{8\pi}{3}
$$

### Resultado Final:
$$
\text{Volume do cone} = \frac{8\pi}{3}
$$



```python
### Implementação em Python:

from sympy import symbols, integrate, pi

# Definindo as variáveis simbólicas
r, theta, z = symbols('r theta z')

# Definindo a função a ser integrada (volume)
f = r

# Calculando a integral interna (em relação a r)
integral_r = integrate(f, (r, 0, z))

# Calculando a integral intermediária (em relação a theta)
integral_theta = integrate(integral_r, (theta, 0, 2*pi))

# Calculando a integral externa (em relação a z)
integral_resultado = integrate(integral_theta, (z, 0, 2))

# Exibindo o resultado
print('Resultado da integral tripla (volume do cone):', integral_resultado)

```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Definindo os limites
theta = np.linspace(0, 2 * np.pi, 100)
z = np.linspace(0, 2, 100)
theta, z = np.meshgrid(theta, z)
r = (z / 2) * 2  # Relação r = (R/h) * z
X = r * np.cos(theta)
Y = r * np.sin(theta)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, z, cmap='viridis', alpha=0.8)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Volume de um cone')

plt.show()
```

Vamos resolver um exemplo de integral tripla para calcular o volume de uma esfera de raio $ R = 1 $ usando coordenadas esféricas.

### Exemplo Teórico: Integral Tripla em Coordenadas Esféricas
**Problema**: Calcule o volume de uma esfera de raio 1.

### Solução Teórica:
Em coordenadas esféricas, as variáveis $ x, y, z $ são representadas por:
- $ x = \rho \sin(\phi) \cos(\theta) $
- $ y = \rho \sin(\phi) \sin(\theta) $
- $ z = \rho \cos(\phi) $

O jacobiano de transformação para coordenadas esféricas é $ \rho^2 \sin(\phi) $, então o elemento de volume é:
$$
dV = \rho^2 \sin(\phi) \, d\rho \, d\phi \, d\theta
$$

Para calcular o volume da esfera, usamos a função constante $ f(\rho, \phi, \theta) = 1 $:
$$
\iiint_V 1 \cdot \rho^2 \sin(\phi) \, d\rho \, d\phi \, d\theta
$$

### Limites de Integração:
- $ \rho $ varia de 0 a 1 (raio da esfera)
- $ \phi $ varia de 0 a $ \pi $ (ângulo polar)
- $ \theta $ varia de 0 a $ 2\pi $ (ângulo azimutal)

### Solução Passo a Passo:
1. **Integral em relação a $ \rho $**:
$$
\int_{0}^{1} \rho^2 \, d\rho = \frac{\rho^3}{3} \Big|_{0}^{1} = \frac{1}{3}
$$

2. **Integral em relação a $ \phi $**:
$$
\int_{0}^{\pi} \sin(\phi) \, d\phi = -\cos(\phi) \Big|_{0}^{\pi} = -\cos(\pi) + \cos(0) = 2
$$

3. **Integral em relação a $ \theta $**:
$$
\int_{0}^{2\pi} 1 \, d\theta = \theta \Big|_{0}^{2\pi} = 2\pi
$$

### Resultado Final:
$$
\iiint_V 1 \cdot \rho^2 \sin(\phi) \, d\rho \, d\phi \, d\theta = \frac{1}{3} \cdot 2 \cdot 2\pi = \frac{4\pi}{3}
$$

O volume da esfera é:
$$
V = \frac{4\pi}{3}
$$


```python

from sympy import symbols, integrate, pi, sin

# Definindo as variáveis simbólicas
rho, phi, theta = symbols('rho phi theta')

# Definindo a função e o jacobiano esférico
f = rho**2 * sin(phi)

# Calculando a integral em relação a rho
integral_rho = integrate(f, (rho, 0, 1))

# Calculando a integral em relação a phi
integral_phi = integrate(integral_rho, (phi, 0, pi))

# Calculando a integral em relação a theta
integral_resultado = integrate(integral_phi, (theta, 0, 2*pi))

# Exibindo o resultado
print('Resultado da integral tripla (volume da esfera):', integral_resultado)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Criando uma grade de pontos no plano xy
X = np.linspace(0, np.pi, 100)
Y = np.linspace(0, np.pi/2, 100)
X, Y = np.meshgrid(X, Y)
Z = np.cos(X) * np.sin(Y)

# Plotando a superfície
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='viridis')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Superfície da função z = cos(x) + sin(y)')

plt.show()
```

```python
import sympy as sp
from sympy import integrate, symbols

# Definindo as variáveis simbólicas
r, theta = symbols('r theta')

# Definindo a função em coordenadas polares
# f = (x + 2y)*Jacobian = r*(cos(theta) + 2*sin(theta)) * r
f = r * (sp.cos(theta) + 2*sp.sin(theta)) * r

# Calculando a integral interna em relação a r
integral_interna = integrate(f, (r, 0, 1))

# Calculando a integral externa em relação a theta
integral_resultado = integrate(integral_interna, (theta, 0, 2*sp.pi))

# Exibindo o resultado
print('Resultado da integral dupla:', integral_resultado)
```

```python
import sympy as sp
from sympy import integrate

# Definindo as variáveis simbólicas
x, y = symbols('x y')

# Definindo a função a ser integrada
f = x**2 - y**2

# Calculando a integral interna (em relação a y)
integral_interna = integrate(f, (y, 0, 2*x))

# Calculando a integral externa (em relação a x)
integral_resultado = integrate(integral_interna, (x, 0, 2))

# Exibindo o resultado
print('Resultado da integral dupla sobre a região triangular:', integral_resultado)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Criando uma grade de pontos no plano xy
X = np.linspace(0, 2, 100)
Y = np.linspace(0, 2*X, 100)
X, Y = np.meshgrid(X, Y)
Z = X**2 - Y**2

# Limitando a região triangular (y <= x)
Z[Y > X] = np.nan  # Exclui valores fora da região

# Plotando a superfície
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='plasma')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('Superfície da função z = xy sobre a região triangular')

plt.show()
```
