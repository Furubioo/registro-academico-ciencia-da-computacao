Questão 1

```python
x = sp.Symbol('x', real=True)
k = sp.Symbol('k', real=True)
k = 2/9
f_poly = k * (3*x-x**2)
area_poly = integrate(f_poly, (x, 0, 3))
prob_poly_1_3 = integrate(f_poly, (x, 1, 3))
media_poly = integrate(x * f_poly, (x, 0, 3))
area_poly, prob_poly_1_3, media_poly
```

Questão 2

```python
x = sp.Symbol('x', real=True)
f_poly = 5 * (1-x)**4
area_poly = integrate(f_poly, (x, 0, 1))
prob_poly_0_02 = integrate(f_poly, (x, 0.2, 0.6))
media_poly = integrate(x * f_poly, (x, 0, 1))
area_poly, prob_poly_0_02, media_poly
```

É uma PDF e a média é 1/6



Questão 3

```python
t = sp.Symbol('t', real=True, positive=True)
lambda_val = 0.1
f_exp = lambda_val * exp(-0.1* t)
area_exp = integrate(f_exp, (t, 0, oo))
prob_exp_0_1 = integrate(f_exp, (t, 0, 5)).evalf()
prob_exp_gt_5 = integrate(f_exp, (t, 8, oo)).evalf()
media_exp = integrate(t * f_exp, (t, 0, oo)).evalf()
area_exp, prob_exp_0_1, prob_exp_gt_5, media_exp
```
